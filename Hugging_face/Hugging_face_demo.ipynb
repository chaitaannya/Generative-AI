Fine Tuning a dataset

```python
from datasets import load_dataset
dataset=load_dataset("stanfordnlp/imdb")
```

```python
dataset
```

preprocessing

```python
from transformers import AutoTokenizer
tokenizer=AutoTokenizer.from_pretrained("bert-base-cased")
def tokenize_function(examples):
  return tokenizer(examples['text'],padding="max_length",truncation=True)

tokenize_datasets=dataset.map(tokenize_function,batched=True)
```

```python
tokenize_datasets
```

```python
tokenize_datasets['train'][0]
```

setting up training arguments

```python
import transformers
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",       # output directory
    eval_strategy="epoch",  # <-- fixed argument name
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01
)

training_args
```

```python
from transformers import AutoModelForSequenceClassification,Trainer
```

```python
model=AutoModelForSequenceClassification.from_pretrained("bert-base-cased",num_labels=2)

trainer=Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenize_datasets["train"],
    eval_dataset=tokenize_datasets["test"],
)
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"

trainer.train()
```

Evaluate the Model

```python
results=trainer.evaluate()
print(results)
```
