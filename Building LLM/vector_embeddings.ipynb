```python
import gensim.downloader as api
model=api.load("word2vec-google-news-300")
```

```python
word_vectors=model
print(word_vectors["computer"])
```

```python
print(word_vectors["cat"].shape)
```

```python
print(word_vectors.most_similar(positive=["Man","boy"], negative=["girl"]))
```

```python
print(word_vectors.similarity('man','boy'))
```

```python
print(word_vectors.similarity('man','woman'))
```

```python
print(word_vectors.similarity('man','cat'))
```

```python
print(word_vectors.similarity('cat','kitten'))
```
